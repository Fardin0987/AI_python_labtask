```python



def get_successors(state):
    
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    successors = []
    
   
    blank_r, blank_c = -1, -1
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                blank_r, blank_c = r, c
                break
                
    
    for dr, dc in moves:
        new_r = blank_r + dr
        new_c = blank_c + dc
        
     
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            
            new_state = [list(row) for row in state]
           
            new_state[blank_r][blank_c] = new_state[new_r][new_c]
            new_state[new_r][new_c] = 0
            
           
            tuple_version = tuple(tuple(row) for row in new_state)
            successors.append(tuple_version)
            
    return successors

def run_bfs(start, goal):
    
    frontier = [(start, [start])]
    explored = set()
    expansion_sequence = []
    
    while len(frontier) > 0:
       
        current_state, path = frontier.pop(0)
        
        if current_state in explored:
            continue
            
      
        expansion_sequence.append(current_state)
        explored.add(current_state)
        
        
        if current_state == goal:
            return expansion_sequence, path, len(path) - 1
            
        
        for neighbor in get_successors(current_state):
            if neighbor not in explored:
                
                already_in_frontier = False
                for state_in_f, _ in frontier:
                    if state_in_f == neighbor:
                        already_in_frontier = True
                        break
                
                if not already_in_frontier:
                    frontier.append((neighbor, path + [neighbor]))
                    
    return None, None, None


initial_state = ((1, 2, 3), (4, 0, 6), (7, 5, 8))
goal_state = ((1, 2, 3), (4, 5, 6), (7, 8, 0))


exp_seq, sol_path, cost = run_bfs(initial_state, goal_state)


print("BFS")
print("Sequence of Node Expansions:", len(exp_seq))
print("Total Path Cost:", cost)
print("\nSolution Path:")
for index, grid in enumerate(sol_path):
    print(f"Step {index}:")
    for row in grid:
        print(row)
    print()
```
