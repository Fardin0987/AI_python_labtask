```python
gbfs_tree = {
    'S': ['A', 'B'], 'A': ['C', 'D'], 'B': ['E', 'F'],
    'C': ['P'], 'D': [], 'E': ['H'], 'F': ['I', 'G'],
    'P': [], 'H': [], 'I': [], 'G': []
}

h_gbfs = {
    'A': 12, 'B': 4, 'C': 7, 'D': 3, 'E': 8, 
    'F': 2, 'H': 4, 'I': 9, 'S': 13, 'P': 20, 'G': 0
}

def gbfs(graph, heuristics, start, goal):
    open_list = [[start]]
    while open_list:
        open_list.sort(key=lambda path: heuristics[path[-1]])
        path = open_list.pop(0)
        node = path[-1]
        
        if node == goal:
            return path
            
        for neighbor in graph[node]:
            if neighbor not in path:
                open_list.append(path + [neighbor])

print("GBFS Path:", gbfs(gbfs_tree, h_gbfs, 'S', 'G'))  
```
