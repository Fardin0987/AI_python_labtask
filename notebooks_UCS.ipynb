```python

def get_successors(state):
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    successors = []
    
    blank_r, blank_c = -1, -1
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                blank_r, blank_c = r, c
                break
                
    for dr, dc in moves:
        new_r = blank_r + dr
        new_c = blank_c + dc
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_state = [list(row) for row in state]
            new_state[blank_r][blank_c] = new_state[new_r][new_c]
            new_state[new_r][new_c] = 0
            successors.append(tuple(tuple(row) for row in new_state))
    return successors

def run_ucs(start, goal):
    frontier = [(0, start, [start])]
    explored = {}
    expansion_sequence = []
    
    while len(frontier) > 0:
        lowest_index = 0
        for i in range(1, len(frontier)):
            if frontier[i][0] < frontier[lowest_index][0]:
                lowest_index = i
                
        current_cost, current_state, path = frontier.pop(lowest_index)
        
        if current_state in explored and explored[current_state] <= current_cost:
            continue
            
        expansion_sequence.append(current_state)
        explored[current_state] = current_cost
        
        if current_state == goal:
            return expansion_sequence, path, current_cost
            
        for neighbor in get_successors(current_state):
            step_cost = 1
            new_cost = current_cost + step_cost
            
            if neighbor not in explored or new_cost < explored[neighbor]:
                frontier.append((new_cost, neighbor, path + [neighbor]))
                
    return None, None, None

initial_state = ((1, 2, 3), (4, 0, 6), (7, 5, 8))
goal_state = ((1, 2, 3), (4, 5, 6), (7, 8, 0))

exp_seq, sol_path, cost = run_ucs(initial_state, goal_state)

print("UCS")
print("Sequence of Node Expansions:", len(exp_seq))
print("Total Path Cost:", cost)
print("\nSolution Path:")
for index, grid in enumerate(sol_path):
    print(f"Step {index}:")
    for row in grid:
        print(row)
    print()
```
