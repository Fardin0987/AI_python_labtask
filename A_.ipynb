```python
astar_graph = {
    'P': [('Q', 1), ('R', 2)],
    'Q': [('T', 12), ('R', 2), ('S', 5)],
    'R': [('S', 2)],
    'S': [('T', 3)],
    'T': []
}

h_astar = {'P': 17, 'Q': 6, 'R': 2, 'S': 1, 'T': 0}

def astar(graph, heuristics, start, goal):
    open_list = [(heuristics[start], 0, [start])]
    while open_list:
        open_list.sort()
        f, g, path = open_list.pop(0)
        node = path[-1]
        
        if node == goal:
            return path
            
        for neighbor, cost in graph[node]:
            if neighbor not in path:
                new_g = g + cost
                new_f = new_g + heuristics[neighbor]
                open_list.append((new_f, new_g, path + [neighbor]))

print("A* Path:", astar(astar_graph, h_astar, 'P', 'T'))
```
